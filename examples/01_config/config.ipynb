# Example Config Import
- Example of how the configuration file is loaded
- Beispiel für eine Fehlermeldung

```python
from encodapy.config import ConfigModel
from IPython.display import display
```

## Example of successful loading of a config file

```python
config = ConfigModel.from_json('config.json')
display(config)

display(config.interfaces.file)

display(config.inputs)

display(config.inputs[0].attributes)

display(config.controller_components)

for controller in config.controller_components:
    display(controller)

display(config.controller_settings.time_settings)
```

## Example of successful loading of a config file

```python
from loguru import logger
```

```python
try:
    config = ConfigModel.from_json('config_error.json')
    display(config)
except Exception as e:
    logger.error(f"Error loading config: {e}")
```

# Example of Input Data Attributes
- Example of how an InputDataAttributeModel can be used
- Use in BasicService

```python
from encodapy.utils.models import (
    InputDataModel,
    InputDataEntityModel,
    OutputDataEntityModel,
    InputDataAttributeModel,
    OutputDataAttributeModel
)
from datetime import datetime, timezone


input_attribute = InputDataAttributeModel(
    id='test_attribute',
    data_type='value',
    data='test',
    data_available=True,
    latest_timestamp_input=datetime.now(timezone.utc),
)

input_entity = InputDataEntityModel(
    id='test_input',
    attributes=[input_attribute],
)
display(input_entity)

output_attribute = OutputDataAttributeModel(
    id='test_attribute',
    latest_timestamp_output=datetime.now(timezone.utc),
)

output_entity = OutputDataEntityModel(
    id='test_output',
    attributes_status=[output_attribute],
)
display(output_entity)

data = InputDataModel(
    input_entities=[input_entity, input_entity],
    output_entities=[output_entity],
    static_entities=[]
)
display(data)
```

```python

data_available = [input_entity_1.attributes for input_entity_1 in data.input_entities]
display(data_available)

data_available = [attribute.data_available for input_entity in data.input_entities
                  for attribute in input_entity.attributes]
display(data_available)
```
